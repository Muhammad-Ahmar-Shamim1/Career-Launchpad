```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

 
```

```python
df = pd.read_csv('../data/CREDIT CARD.csv')
```

```python
# 1. Class Distribution
print("Class Distribution:")
print(df['Class'].value_counts(normalize=True))
```

```python
# 2. Visualize Fraud vs Non-Fraud
sns.countplot(x='Class', data=df)
plt.title('Transaction Class Distribution (0: Normal, 1: Fraud')
```

```python
# 3. Amount Analysis
plt.figure(figsize=(10, 6))
sns.boxplot(x='Class', y='Amount', data=df)
plt.title('Transaction Amount: Normal vs Fraud')
plt.yscale('log') # Log scale to see variations better
```

```python
plt.figure(figsize=(6, 5))
sns.heatmap(df.corr(), cmap='coolwarm', annot=False)
plt.title('Feature Correlation Heatmap')
```
